```python
import pandas as pd
import numpy as np
```

```python
# 1.
orders = pd.DataFrame({'cust': ['A', 'A', 'B', 'C', 'C', 'C'],
'amount': [10, 20, 15, 5, 5, 30]})
```

```python
high= orders.groupby('cust')['amount'].sum().sort_values(ascending=False)
```

```python
total_count=orders['cust'].nunique()
repurchase = (orders['cust'].value_counts()>1).sum()
total_count
repurchase
```

```python
print(f'재구매 비율 : {repurchase/total_count*100:.1f}%')
print(f'최고금액고객 : {high.index[0]}')
```

```python
# 2. 
students = pd.DataFrame({
'id': ['S01', 'S02', 'S03', 'S04'],
'name': ['김민수', '이서연', '박지훈', '최유진']
})

scores = pd.DataFrame({
'id': ['S01', 'S02', 'S01', 'S03', 'S04', 'S02'],
'subject': ['수학', '수학', '영어', '영어', '수학', '영어'],
'score': [85, 90, 95, 80, 75, 88]
})
```

```python
mix = pd.merge(students, scores, on='id')
```

```python
sortmix = mix.groupby('name')['score'].mean().sort_values(ascending=False)
sortmix
```

```python
scores_mean = mix['score'].mean()
scores_mean
```

```python
high_score = mix.loc[mix['score'].idxmax()]
```

```python
print(sortmix)
print(scores_mean)
print(high_score['name'], high_score['score'])
```
