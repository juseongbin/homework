```python
from bs4 import BeautifulSoup

def check(html, selector):
    found = BeautifulSoup(html, 'html.parser').select(selector)
    print(f"{selector} -> {len(found)}개")
    for el in found[:5]:
        print("    ", el.name, "|", el.get_text(" ", strip=True)[:40])
    return len(found)
```

```python
import requests

url = "https://books.toscrape.com/"
res = requests.get(url)
res.raise_for_status()
text = res.text
```

```python
# 상품 반복 단위
check(text, 'div.image_container')
# 상품명
check(text, 'h3')
# 가격
check(text, 'p.price_color')
# 별점
check(text, 'p.star-rating')
# 재고상태
check(text, 'i.icon-ok')
```

```python
# 상품 상세 링크
check(text, 'article.product_pod div.image_container a')
# 상품 카드 내부 전체 링크
check(text, 'article.product_pod a')
# 다음 페이지
check(text, 'li.next a')
# 카테고리 링크
check(text, 'div.side_categories ul li a')
```

```python
check(text, 'ul.nav.nav-list ul li')
```
