```python
import pandas as pd
import csv
```

```python
url = 'raw_sales_data.csv'
df = pd.read_csv(url, thousands=',', na_values=['-'])
df.head()
```

```python
#데이터의 행과 열 개수 확인
print(f'데이터출력 : {df.shape}')
```

```python
#열 이름 확인
print(f'열 이름 : \n{df.columns}')
```

```python
#상위 5개 데이터 확인
print(f'상위 5개 데이터 : \n{df.head()}')
```

```python
#각 열의 데이터 타입 확인
print(f'데이터 타입 : \n{df.dtypes}')
```

```python
#전체 데이터의 정보 확인
print('데이터 정보:')
print(df.info())
```

```python
#수치형 데이터의 형균, 최솟값, 최댓값 등 통계 확인
print('수치형 데이터 통계:')
print(df.describe())
```

```python
#전체 행을 기준으로 중복 데이터 개수 확인
#OrderID를 기준으로 중복된 주문 개수 확인
print('=====2. 중복 데이터 확인=====')
print(f'전체 중복 행 : {df.duplicated().sum()}')
print(f'중복 주문번호 : {df['OrderID'].duplicated().sum()}')
```

```python
#중복된 주문 데이터 출력
#OrderID가 중복된경우 첫번째 데이터만 남기고 제거
print('중복 주문 데이터:')
df[df['OrderID'].duplicated()]
##답이랑 다르긴 한데 일단 그냥 중복 제거함
##아마 중복되는 첫번째애들이 안나오는듯
```

```python
#중복 제거 후 데이터 개수 확인
df = df.drop_duplicates(subset='OrderID')
print(f'중복 제거 후 데이터 크기 : {df.shape}')
```

```python
#3. 결측치 확인
#열별 결측치 개수 확인
#결측치가 존재하는 열만 출력
print('===== 3. 결측치 확인 =====')
print('결측치가 존재하는 열:')
missing = df.isna().sum()
print(missing[missing>0])
print()
print(f'전체 결측치: {df.isna().sum().sum()}')
```

```python
##Quantity 음수 결측치로
##df.loc[df['Quantity'] <= 0, 'Quantity'] = NaN
##unitprice 숫자형 변환
df['UnitPrice'] = df['UnitPrice'].astype(float)
```

```python
print('=====4. 데이터 타입 및 잘못된 값 처리 =====')
print('변환 후 데이터 타입:')
print(df[['Quantity', 'UnitPrice']].dtypes)
```

```python
#4. 결측치 처리
df.drop(df[df["OrderID"].isna()].index, axis=0, inplace=True)
df.drop(df[df["OrderDate"].isna()].index, axis=0, inplace=True)
df['City'] = df['City'].fillna(df['City'].mode()[0])
df['Category'] = df['Category'].fillna(df['Category'].mode()[0])
df['PaymentMode'] = df['PaymentMode'].fillna(df['PaymentMode'].mode()[0])
df['Quantity'] = df['Quantity'].fillna(df['Quantity'].median())
df['UnitPrice'] = df['UnitPrice'].fillna(df['UnitPrice'].median())
```

```python
print('===== 5. 결측치 처리 =====')
```

```python
print('===== 6. 최종 데이터 확인 =====')
```

```python
print(f'최종 데이터 크기 : {df.shape}')
print(f'중복 주문번호 : {df['OrderID'].duplicated().sum()}')
print()
print(f'열별 결측치 : \n{df.isna().sum()}\n')
print(f'전체 결측치 : {df.isna().sum().sum()}\n')
print(f'최종 데이터 타입 : \n{df.dtypes}\n')
```

```python
print(f'최종 데이터 통계 : \n{df.describe()}')
```

```python
df.to_csv('cleaned_sales_data.csv', index=False)
print('정제된 데이터 저장 완료!')
print('파일명 : cleaned_sales_data.csv')
```
