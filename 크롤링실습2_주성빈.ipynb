```python
from bs4 import BeautifulSoup
import requests, time, re
```

```python
HEADERS = {'User-Agent' : 'Mozilla/5.0 (Windows NT 10.0; Win64; x64) AppleWebKit/537.36 (KHTML, like Gecko) Chrome/158.0.0.0 Safari/537.36 Edg/158.0.0.0'}
url = 'https://news.naver.com/main/list.naver?mode=LSD&mid=sec&sid1=001&date=20261125&page=1'
res = requests.get(url, headers=HEADERS)
```

```python
count = 0
for i in range(5):
    url = f'https://news.naver.com/main/list.naver?mode=LSD&mid=sec&sid1=001&date=20261125&page={i+1}'
    res = requests.get(url, headers=HEADERS)    

    if res.status_code == 200:
        html = res.text
        soup = BeautifulSoup(html, 'html.parser')
        headlines = soup.select("div.list_body .type06_headline li dl dt:not(.photo) > a")
        link = headline.get('href')
        res2 = requests.get(link, headers=HEADERS)
        html2 = res2.text
        soup2 = BeautifulSoup(html2, 'html.parser')
        script = soup2.select("#dic_area:not(#end_photo_org)")
        for i, headline in enumerate(headlines, 1):
            print(10*count+i, headline.get_text(strip=True))

    else:
        print(res.status_code)
    count += 1
```
