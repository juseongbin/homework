```python
from bs4 import BeautifulSoup
import requests, time
```

```python
HEADERS = {'User-Agent' : 'Mozilla/5.0'}
def fetch(url):
    res = requests.get(url, headers = HEADERS, timeout=10)
    res.raise_for_status()
    return(res)
```

```python
res = fetch('https://quotes.toscrape.com/')
soup = BeautifulSoup(res.text, 'html.parser')
```

```python
author = soup.select('small.author')

for i in author:
    print(i.get_text())
```

```python
# find_all 수정
authors = soup.find_all('small', class_='author')

for i in authors:
    print(i.get_text())
```

```python
html = """

<html>

<head>

<title>작품과 작가 모음</title>

</head>

<body>

<h1>책 정보</h1>

<p id="book_title">토지</p>

<p id="author">박경리</p>


<p id="book_title">태백산맥</p>

<p id="author">조정래</p>

<p id="book_title">감옥으로부터의 사색</p>

<p id="author">신영복</p>

</body>

</html>

"""
```

```python
soup = BeautifulSoup(html, 'html.parser')
```

```python
title = soup.find_all('p', id='book_title')
for i in title:
    print(i.get_text())
```

```python
author = soup.find_all('p', id='author')
for i in author:
    print(i.get_text())
```

```python
for i, j in zip(title, author):
    print(f'{i.get_text()}/{j.get_text()}')
```
